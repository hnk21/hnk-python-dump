### Breadth First Search (2D Array)

```python
from collections import deque as queue
 
# up down right left
directions = [ [-1, 0], [1, 0], [0, -1], [0, 1]]
 
def BFS(grid, visited, row, col):
    row_bounds, col_bounds = len(grid), len(grid[0])

    traverse_queue = queue()
    traverse_queue.append((row, col))
 
    # Mark the starting position as visited and push it into the queue
    visited[row][col] = True
    
    # Function to check if position is be visited or not
    def toVisit(visited, row, col):
        if (row < 0 or row >= row_bounds or col < 0 or col >= col_bounds):
            return False
        if (visited[row][col]):
            return False
        return True

    # Iterate while the queue is not empty
    while (len(traverse_queue) > 0):
        pos = traverse_queue.popleft()
        row, col = pos[0], pos[1]

        print(grid[row][col], end = " > ")

        # Move to adjacent positions
        for move_row, move_col in directions:
            next_row = row + move_row
            next_col = col + move_col

            if (toVisit(visited, next_row, next_col)):
                traverse_queue.append((next_row, next_col))
                visited[next_row][next_col] = True
```

### Breadth First Search (Binary Tree)

```python
def BFS_BST(root):
    nodes = []
    queue = [root]

    while queue:
        size = len(queue)
        for i in range(0, size):
            current = queue.pop(0)

            nodes.append(current.val)

            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)

    return nodes
```

### Depth First Search (Graph)

```python
from collections import defaultdict
 
class Graph:
    def __init__(self):
        self.graph = defaultdict(list)
 
    def addEdge(self, u, v):
        self.graph[u].append(v)

    # DFS
    def DFSUtil(self, v, visited):
        # Mark the current node as visited and print it
        visited.add(v)
        print(v, end=' ')
 
        # Recur for all the vertices adjacent to this vertex
        for neighbour in self.graph[v]:
            if neighbour not in visited:
                self.DFSUtil(neighbour, visited)
 
    # The function to do DFS traversal. It uses recursive DFSUtil()
    def DFS(self, v):
        visited = set() # Create a set to store visited vertices
        self.DFSUtil(v, visited)
```

### Depth First Search (2D Array)

```python
ROW = 3
COL = 3
 
# Direction vectors
dRow = [0, 1, 0, -1]
dCol = [-1, 0, 1, 0]
vis = [[False for i in range(3)] for j in range(3)]
 
def isValid(row, col):
    global ROW
    global COL
    global vis

    if (row < 0 or col < 0 or row >= ROW or col >= COL): return False
    if (vis[row][col]): return False
    return True
 

def DFS(row, col, grid):
    global dRow
    global dCol
    global vis
     
    st = []
    st.append([row, col])
 

    while (len(st) > 0):
        curr = st[len(st) - 1]
        st.remove(st[len(st) - 1])
        row = curr[0]
        col = curr[1]
 

        if (isValid(row, col) == False): continue
 
        # Mark the current cell as visited
        vis[row][col] = True
 
        print(grid[row][col], end = " ")
 
        for i in range(4):
            adjx = row + dRow[i]
            adjy = col + dCol[i]
            st.append([adjx, adjy])
```

### Topological Sort (DFS)

```python
from collections import defaultdict

class Graph:
    def __init__(self, vertices):
        self.graph = defaultdict(list)
        self.V = vertices
 
    def addEdge(self, u, v):
        self.graph[u].append(v)
 
    
    def topologicalSortUtil(self, v, visited, stack):
        visited[v] = True

        for i in self.graph[v]:
            if visited[i] == False:
                self.topologicalSortUtil(i, visited, stack)
 
        stack.append(v)
 
    def topologicalSort(self):
        visited = [False]*self.V
        stack = []
 
        for i in range(self.V):
            if visited[i] == False:
                self.topologicalSortUtil(i, visited, stack)
 
        print(stack[::-1])  # return list in reverse order
```

### Insertion Sort

```python
def insertionSort(a):
  for i in range(len(a)):
    key = a[i]
    j = i-1
    while j > -1 and a[j] > key:
        a[j+1] = a[j]
        j = j-1
    a[j+1] = key
  
  return a

array = [2,8,5,3,9,4,1,7]
insertionSort(array)
```

```python
def isSubsequence(s: str, t: str) -> bool:
    if len(s) > len(t): return False
    p1, p2 = 0, 0
    while p1 < len(s) and p2 < len(t):
        if s[p1] == t[p2]:
            p1 += 1
            p2 += 1
        else:
            p2 += 1
    return p1 >= len(s)
```

```python
def maxArea(height: list[int]) -> int:
    max_volume = 0
    left, right = 0, len(height) - 1
    while left != right:
        max_volume = max(max_volume, (right-left) * min(height[left], height[right]))
        # print(f"> {left}, {right} -> {max_volume}")
        
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    
    return max_volume
```

# Sliding Window

```python
def longestOnes(nums, k):        
    left = 0
    zeroCount = 0
    for right in nums:
        if right == 0:
            zeroCount += 1
        if zeroCount > k:
            if nums[left] == 0:
                zeroCount -= 1
            left += 1
    return len(nums) - left
```

```python
def findMaxAverage(nums: list[int], k: int) -> float:
    window_sum = sum(nums[:k])
    max_value = window_sum
    for i in range (k, len(nums)):
        window_sum = window_sum - nums[i-k]+ nums[i]
        if window_sum >= max_value:
            max_value = window_sum 
    return max_value/k
```

```python
def maxVowels(s: str, k: int) -> int:
    if len(s) < k:
        return 0
    vowels = set('aeiou')
    vowels_in_window = sum(s[:k].count(v) for v in vowels)
    max_ = vowels_in_window
    for i in range(k, len(s)):
        if s[i - k] in vowels:
            vowels_in_window -= 1
        if s[i] in vowels:
            vowels_in_window += 1
            if vowels_in_window > max_:
                max_ = vowels_in_window
                if max_ == k:
                    return k
    return max_
```

### Monotonic Stack (Increasing Order)

Maintains a stack that is strictly increasing, given an unsorted input array

```python
def monotonic_increasing_stack(arr):
    stack = []
    
    for num in arr:
        while stack and stack[-1] > num:
            stack.pop()
        stack.append(num)
    
    return stack

arr = [2, 1, 2, 4, 3, 1, 6]
print(monotonic_increasing_stack(arr))
arr = [2, 1, 2, 4, 3, 6]
print(monotonic_increasing_stack(arr))
```

```python
from collections import Counter

def removeDuplicateLetters(self, s: str) -> str:
    letters = Counter(s)
    stack = []

    for c in s:
        print(stack, " >>>", end = " ")
        letters[c] -= 1
        if c in stack:
            continue
        while stack and stack[-1] > c and letters[stack[-1]] > 0:
            stack.pop()
        stack.append(c)
        print(stack)
    return "".join(stack)
```

```python
class StockSpanner:
    def __init__(self):
        self.stack = []

    def next(self, price: int) -> int:
        span = 1
        
        while self.stack and self.stack[-1][0] <= price:
            span += self.stack[-1][1]
            self.stack.pop()
        
        self.stack.append((price, span))
        
        return span
```

```python
def compute_prefix_sum(arr):
    n = len(arr)
    prefix_sum = [0] * n
    prefix_sum[0] = arr[0]

    for i in range(1, n):
        print(f"{prefix_sum[i - 1]} > {prefix_sum[i - 1] + arr[i]}")
        prefix_sum[i] = prefix_sum[i - 1] + arr[i]
    
    return prefix_sum
```

### N-Queens

https://www.geeksforgeeks.org/dsa/n-queen-problem-backtracking-3/

Naive Backtracking

- Place a Queen on a row
- Check if current position is safe
    - If safe - OK - Go next column and try from starting row
    - If not safe - backtrack and try other position

```python
def nQueens(n):
    board = [[0] * n for _ in range(n)]
    result = []
    putQueens(row = 0, board = board, result = result)
    return result

def putQueens(row, board, result):
    n = len(board)

    # Base case: All Queens are placed
    if row == n:
        ans = []
        for i in range(n):
            for j in range(n):
                if board[i][j]:
                    ans.append(j + 1)
        
        result.append(ans)
        return
    
    # Check each position on current 'row' 
    for i in range(n):
        if isSafe(board, row, i):
            board[row][i] = 1
            
            # Recurse and check next row on the board
            putQueens(row + 1, board, result)

            # Backtrack
            board[row][i] = 0

def isSafe(board, row, col):
    n = len(board)

    # Check column
    for i in range(n):
        if board[i][col]:
            return False
    
    # No need to check lower diags since Queens are populated from top row downwards
    # Check upper diagonal, left side
    i, j = row-1, col-1
    while i >= 0 and j >= 0:
        if board[i][j]:
            return False
        i -= 1
        j -= 1

    # Check upper diagonal, right side
    i, j = row-1, col+1
    while i >= 0 and j < n:
        if board[i][j]:
            return 0
        i -= 1
        j += 1

    return True
```

```python
import time
n = 14

start = time.time()
result = nQueens(n)
end = time.time()

print(f"Naive Backtracking took {round(end-start, 2)} seconds to run.")
print(f"For {n} Queens on a {n}x{n} board, there are {len(result)} possible placements.")
```

Backtracking with Hashing

- Use 3 arrays to track occupied columns and diagonals

```python
def nQueens_hashing(n):
    cols = [0] * n
    leftDiag, rightDiag = [0] * (2 * n), [0] * (2 * n)
    curr, result = [], []
    putQueens_hashing(0, cols, leftDiag, rightDiag, curr, result)
    return result

def putQueens_hashing(i, cols, leftDiag, rightDiag, curr, result):
    n = len(cols)

    if i == n:
        result.append(curr[:])
        return
    
    for j in range(n):
        # Check if Queen can be placed
        if cols[j] or rightDiag[i + j] or leftDiag[i - j + n - 1]:
            continue

        # Mark the cell occupied
        cols[j] = 1
        rightDiag[i + j] = 1
        leftDiag[i - j + n - 1] = 1
        curr.append(j + 1)

        putQueens_hashing(i + 1, cols, leftDiag, rightDiag, curr, result)

        # Remove Queen from current cell
        curr.pop()
        cols[j] = 0
        rightDiag[i + j] = 0
        leftDiag[i - j + n - 1] = 0
```

```python
import time
n = 14

start = time.time()
result = nQueens_hashing(n)
end = time.time()

print(f"Backtracking with Hashing took {round(end-start, 2)} seconds to run.")
print(f"For {n} Queens on a {n}x{n} board, there are {len(result)} possible placements.")
```

### Get Maximum Depth

```python
def maxDepth(self, root: Optional[TreeNode]) -> int:
    if not root:
        return 0
    
    return 1 + max(self.maxDepth(root.left), self.maxDepth(root.right))
```
