### defaultdict

```python
from collections import defaultdict
words = ["apple", "banana", "apple", "orange", "banana", "apple"]
word_count = defaultdict(int)

for word in words:
  word_count[word] += 1

print(dict(word_count))
```

> Counter() dictionary always arranges and updates keys starting from the **key with the largest value**

```python
from collections import Counter

tasks = ["A","A","A","A","A","A","B","C","D","E","F","G","B","B","B","B","B","B","B"]
task_remain = Counter(tasks)
print(task_remain, "\n")

task_remain["B"] -= 5
print(task_remain)
```

```python
# elements
c = Counter(a=4, b=2, c=5, d=-2, e=-100)
sorted(c.elements())
```

```python
# most_common
Counter('abracadabra').most_common(3)
```

```python
# subtract
c = Counter(a = 4, b = 2, c = 0, d = -2)
d = Counter(a = 1, b = 2, c = 3, d = 4)
c.subtract(d)
c
```

```python
# total
c = Counter(a = 10, b = 5, c = 0)
c.total()
```

### deque
double ended queue

```python
from collections import deque
queue = deque(['name', 'age', 'DOB'])
print(queue)

# Append
queue.append(4)
print(queue)
queue.appendleft(6)
print(queue)

# Pop
print(f"> {queue}")
x = queue.pop()
print(f"Popped {x}, result > {queue}")
y = queue.popleft()
print(f"Popped {y}, result > {queue}")
```

```python
de = deque([1, 2, 3, 3, 69, 2, 4])

print("Using index() to print the index of the first occurrence of 4")
print (de.index(4))
print("")

print(de)
print("Using insert() to insert the value 3 at index 4")
de.insert(4, 3)
print(de)
print("")

print("Using count() to count the occurrences of 3")
print(de.count(3))
print("")

print(de)
print("Using remove() to remove the first occurrence of 2")
de.remove(2)
print(de)
```

```python
de = deque([1, 2, 3,])
print(de)
print("")

# Using extend() to add numbers to right end adds 4,5,6 to right end
de.extend([4, 5, 6])
print(de)
print("")

# Using extendleft() to add numbers to left end adds 7,8,9 to left end
de.extendleft([7, 8, 9])
print(de)
print("")

# Using rotate() to rotate the deque rotates by 3 to left
de.rotate(-3)
print(de)
print("")

# Using reverse() to reverse the deque
de.reverse()
print (de)
```

### OrderedDict

```python
from collections import OrderedDict
  
od = OrderedDict() 
od['a'] = 1
od['b'] = 2
od['c'] = 3
od['d'] = 4
  
for key, value in od.items(): 
    print(key, value)
```

# dataclasses
https://docs.python.org/3/library/dataclasses.html

```python
# Standard in-built way of declaring classes

class Person:
    def __init__(self, name: str, age: int, job: str):
        self.name = name
        self.age = age
        self.job = job

    def __repr__(self):
        return f"Person(name={self.name!r}, age={self.age!r}, job={self.job!r})"

    def __eq__(self, other):
        if not isinstance(other, Person):
            return NotImplemented
        
        return (self.name == other.name and self.age == other.age and self.job == other.job)
```

```python
# Using dataclasses

from dataclasses import dataclass

@dataclass
class Person:
  name: str
  age: int
  job: str

personA = Person(name="Alice", age=30, job="Engineer")
print(personA)

personB = Person(name="Alice", age=30, job="Engineer")
print(personA == personB)
```

### cycle

```python
from itertools import cycle

count = 0

for i in cycle("ABC"):
  if count > 7:
    break
  else:
    print(i+str(count), end = " ")
    count += 1
```

```python
import itertools

l = ['alpha', 'beta', 'cunt']

iterators = itertools.cycle(l)

for i in range(6):
    print(next(iterators), end=" ")
```

### count

```python
from itertools import count

for i in count(5, 5):
  print(i, end = " ")
  if i == 35:
    break
```

### chain
Effective use of memory

```python
from itertools import chain

list1 = [1,2,3]
list2 = ["a","b","c"]
combined_add = list1 + list2
combined_chain_list = list(chain(list1, list2))
print(combined_add)
print(combined_chain_list)

combined_chain = chain(list1, list2)
print(combined_chain)
print(next(combined_chain))
print(next(combined_chain))
```

# timeit

```python
import timeit

code1 = """
a = [1,2,3,4,5]
b = [x * 2 for x in a]
"""

def code2():
  a = [1,2,3,4,5]
  b = list(map(lambda x: x * 2, a))

time1 = timeit.timeit(code1, number=10000)
time2 = timeit.timeit(code2, number=10000)

print(f"code1 took: {time1}")
print(f"code2 took: {time2}")
```

# pickle
Save any python object into a file

```python
import pickle

class Dog:
    def __init__(self, name, breed, age):
      self.name = name
      self.breed = breed
      self.age = age

    def bark(self):
      print(f"My name is {self.name} and I am a {self.age} year old {self.breed}")

data = Dog("Max", "Golden Retriever", 10)

# Save Dog class 'data' as a .pkl
with open("data.pkl", "wb") as f:
  pickle.dump(data, f)

with open("data.pkl", "rb") as f:
  loaded_data = pickle.load(f)

print(loaded_data)
loaded_data.bark()
```
