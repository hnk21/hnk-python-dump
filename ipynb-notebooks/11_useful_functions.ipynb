### lambda + sorted

#### Sort strings in an array based on their length

```python
words = ["apple", "pie", "banana", "cherry"]
words_sorted_length = sorted(words, key = lambda x: len(x))

print(words_sorted_length)
print( list((word, len(word)) for word in words_sorted_length) )
```

#### Sort dictionary based on value, then get the first k keys

```python
numCount = {"score_1": 7, 
            "score_2": 3, 
            "score_3": 1,
            "score_4": 5, 
            "score_5": 2, 
            "score_6": 1, 
            "score_7": 11}

k = 2

res = sorted( list(numCount.items()), key = lambda x: x[1], reverse = True)[0:k]
print(res)
```

```python
students = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
students_sort_age = sorted(students, key = lambda x: x[1])

print(students)
print(students_sort_age)
```

#### Sort nested array in order of one value of an index, then the other

```python
h = [('a', 5), ('f', 3), ('b', 3), ('c', 3), ('d', 1), ('e', 4)]

sorted_list = sorted(h, key = lambda k: (-k[1], k[0]))
print(sorted_list)
```

### List Comprehension

```python
num_list = [1, 2, 4, 1, 12, 20, 11]
[x // 2 for x in num_list]
```

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x for x in fruits if "a" in x]
print(newlist)
```

### Remove non-alphanumeric parts in a string

```python
string_value = "alphanumeric@123__"
s = ''.join(ch for ch in string_value if ch.isalnum())
print(s)
```

```python
string_value = "alphanumeric@ 1 23__"
s = ''.join(filter(str.isalnum, string_value))
print(s)
```

### Get longest common prefix of array of strings

```python
def longestCommonPrefix(strs: list):
    prefix = min(strs, key=len)
    print(f"prefix > {prefix}")
    for word in strs:
        while prefix != word[:len(prefix)]:
            prefix = prefix[:-1]
    return prefix
```

```python
strs = ["class", "clit", "clooney"]
longestCommonPrefix(strs)
```

### Quick way to get entire english alphabet in an array

```python
alphabet_list = list(map(chr, range(65, 91)))
print(alphabet_list)
```
