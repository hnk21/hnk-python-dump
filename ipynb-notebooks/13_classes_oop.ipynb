# Traditional class setup

https://www.w3schools.com/python/python_classes.asp

https://realpython.com/python-classes/

```python
class Person:
    # Class properties
    species = "Human"

    # __init__ method - initial assignment/operations upon creation of object
    # 'self' parameter - reference to the current instance of the class
    def __init__(self, name: str, age = 10, country = "Singapore"):
        self.name    = name
        self.age     = age
        self.country = country

    # __str__ method: Returned if the class-object is printed directly
    def __str__(self):
        return f"From __str__: {self.name}, {self.age}, {self.country}"

    def print_info(self):
        print(f"This person is {self.species}, with name '{self.name}', age '{self.age}', lives in '{self.country}'.\n")

    def celebrate_birthday(self):
        self.age += 1
        print(f"Happy birthday {self.name}! You are now {self.age}\n")
```

```python
# Creating objects from class
person_1 = Person("Gustave", 33, "Lumiere")
person_1.print_info()

person_2 = Person("Bob")
person_2.print_info()

# Access properties of the object
print(f"Directly printing the object properties: {person_2.name}, {person_2.age}\n")

# Modify properties of the object
person_2.name = "Bobby"
person_2.print_info()

# Call Methods that update properties of the object
person_2.celebrate_birthday()
person_2.celebrate_birthday()
person_2.celebrate_birthday()
```

```python
# Delete properties
del person_2.name
# person_2.print_info() # This will return an AttributeError

# Add new properties
person_2.gender = "Male"
print(person_2.gender)
```

```python
# Printing the object will always call __str__
print(person_2)
```

```python
# self does not have to be named "self"
class Fruit:
    def __init__(this, name, colour, number=0):
        this.name = name
        this.colour = colour
        this.number = number
    
    def fruit_info(this):
        print(f"This is the '{this.name}' fruit, with colour '{this.colour}', count '{this.number}'")

    def print_message(this):
        this.fruit_info() # call method using 'self' (this)
    
    def start(this):
        print("> Enter the number of fruits you have:")
        this.number = input()
        this.print_message()
```

```python
fruit_1 = Fruit("apple", "red")
fruit_1.start()
```

## Inheritance

```python
# Parent Class
class Person:
    def __init__(self, name: str, age: int, sex: str):
        self.name = name
        self.age = age
        self.sex = sex

    def info(self):
        print(f"'{self.name}' is {self.age} years old, and is {self.sex}.\n")

```

```python
# Child Class - without __init__ - Will automatically inherit the parent's __init__
class Student(Person):
    pass
    # The following also keeps the parent's inheritance
    # def __init__(self, name, age, sex):
    #     Person.__init__(self, name, age, sex)

student_1 = Student("Charlie", 15, "Male")
student_1.info()
```

```python
class Student(Person):
    # Adding __init__ prevents the child class from inheriting the parent's __init__
    def __init__(self, name: str, age: int, sex: str, grade: int):
        self.name = name
        self.age = age
        self.sex = sex
        self.grade = grade

student_1 = Student("Charlie", 15, "Male", 4)
student_1.info()
```

```python
# 'super' makes the child class inherit all methods and properties from the parent
class Student(Person):
    def __init__(self, name, age, sex):
        super().__init__(name, age, sex)
```

# dataclasses library

```python
from dataclasses import dataclass

@dataclass
class Person:
    name: str
    age: int
    job: str

person_a = Person(name="Alice", age=30, job="Engineer")
person_b = Person(name="Alice", age=30, job="Engineer")
person_c = Person(name="Alice", age=31, job="Teacher")

print(f"{person_a}\n")
print(f"Is person_a the same as person_b? - {person_a == person_b}\n")
print(f"What about person_b and person_c? - {person_b == person_c}\n")
```
